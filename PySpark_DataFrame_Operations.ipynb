# GROUP BY OPERATIONS

```python
spark = SparkSession.builder.appName("Groupby").getOrCreate()
```

```python
# 1. Group by department and count employees in each department.

df.groupBy("department").count().show()
```

```python
# 2. Find total salary per department.

df.groupBy("department").sum("salary").show()
```

```python
# 3. Find average salary per department.

df.groupBy("department").avg("salary").show()
```

```python
# 4. Group by city and find max salary.
#Importing max function

from pyspark.sql.functions import max

df.groupBy("city").agg(max("salary")).show()
```

```python
# 5. Group by department and gender and count employees.

df.groupBy("department","gender").count().show()
```

# MULTIPLE AGGREGATION

```python
from pyspark.sql.functions import sum,avg,count,max,min,countDistinct
```

```python
# 1. Perform multiple aggregations on salary per department:
#sum
#average
#maximum
#minimum

# 2. Rename all aggregated columns appropriately.

df.groupBy("department").agg(
    sum("salary").alias("Total_Salary"),
    avg("salary").alias("Average_Salary"),
    max("salary").alias("Max_Salary"),
    min("salary").alias("Min_Salary"),
).show()
```

```python
# 3. Group by department and role, then calculate:
# total salary
# average salary
# employee count

df.groupBy("department","role").agg(
    sum("salary").alias("Total_Salary"),
    avg("salary").alias("Average_Salary"),
    count("employee_id").alias("Employee_Count")
).show()




```

```python
# 4. Count distinct roles in each department.

df.groupBy("department").agg(
    countDistinct("role").alias("Distinct_Roles")
).show()
```

```python
# 5. Find the top 3 departments with the highest total salary.

df.groupBy("department").agg(
    sum("salary").alias("Total_Salary")
).orderBy("Total_Salary",ascending=False).limit(3).show()
```
