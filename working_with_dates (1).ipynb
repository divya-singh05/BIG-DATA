```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

# Create Spark Session
spark = SparkSession.builder.appName("StudentDateFunctions").getOrCreate()

# Student Admission Data
data = [
    ("2023-06-15", "20/06/2023"),
    ("2024-01-10", "15/01/2024"),
    ("2024-08-05", "12/08/2024")
]

# Create DataFrame
df = spark.createDataFrame(data, ["Admission_Date", "Verification_Date"])

# Apply Date Functions
result_df = df.select(

    current_date().alias("Current_Date"),

    to_date(col("Admission_Date")).alias("Admission_Date"),

    to_date(col("Verification_Date"), "dd/MM/yyyy").alias("Verification_Date"),

    date_format(col("Admission_Date"), "dd-MM-yyyy").alias("Formatted_Date"),

    add_months(col("Admission_Date"), 3).alias("After_3_Months"),

    date_add(col("Admission_Date"), 15).alias("After_15_Days"),

    datediff(
        to_date(col("Verification_Date"), "dd/MM/yyyy"),
        to_date(col("Admission_Date"))
    ).alias("Days_Between")
)

result_df.show(truncate=False)
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

# Create Spark Session
spark = SparkSession.builder \
    .appName("ProjectDateAnalysis") \
    .getOrCreate()

# Project Data
data = [
    (501, "AI Chatbot", "2022-02-14"),
    (502, "Sales Dashboard", "2023-07-08"),
    (503, "Stock Prediction", "2021-11-20"),
    (504, "Face Detection", "2024-01-05")
]

# Create DataFrame
df = spark.createDataFrame(data, ["Project_ID", "Project_Name", "Start_Date"])

# Convert Start_Date to DateType
df = df.withColumn(
    "Start_Date",
    to_date(col("Start_Date"), "yyyy-MM-dd")
)

# Apply Date Functions
result_df = df.select(

    col("Project_ID"),
    col("Project_Name"),
    col("Start_Date"),

    round(
        months_between(current_date(), col("Start_Date")),
        2
    ).alias("Project_Duration_Months"),

    next_day(col("Start_Date"), "Monday").alias("Next_Monday"),

    trunc(col("Start_Date"), "year").alias("Year_Start"),

    trunc(col("Start_Date"), "month").alias("Month_Start"),

    year(col("Start_Date")).alias("Start_Year"),

    quarter(col("Start_Date")).alias("Quarter"),

    month(col("Start_Date")).alias("Month"),

    dayofweek(col("Start_Date")).alias("Day_Of_Week")
)

result_df.show(truncate=False)
```
