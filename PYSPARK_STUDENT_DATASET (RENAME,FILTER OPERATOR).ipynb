SELECTING ANY 2 COLUMNS

```python
df.select(['Name','Age']).show()
```

ADDING VALUES IN SCORE COLUMN

```python
df_revised = df.withColumn("Marks", df['Marks'].cast('float')-10)
df_revised.show()
```

DROPPING COLUMN

```python
df_final_columns = df_revised.drop('Grade')
df_final_columns.show()
```

RENAMING

```python
df_final = df_final_columns.withColumnRenamed("Marks", "Total_marks")
df_final.show(5)
```

```python
df_final.createOrReplaceTempView("student_data")
df_sql = spark.sql("select * from student_data")
df_sql.show()
```

APPLYING FILTER

```python
df_final.filter("Total_marks>80").show()
```

```python
df_final.filter(df_final["Total_marks"] >85).select(['Name','Age']).show()
```

MULTIPLE FILTER

```python
df_final.filter((df_final["Total_marks"]>85) | (df_final["Total_marks"] <90)).show()
```

```python
df_final.filter(df_final["Total_marks"] != 90).show()
```

```python
df_final.filter(~(df_final['Total_marks']>85) & (df_final['Total_marks'] <90)).show()
```

```python
df_finalmarks = df_final.filter(df_final["Total_marks"]>79).collect()
```
